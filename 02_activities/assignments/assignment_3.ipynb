```python
# Import standard libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

Use scikit-learn to load the Wine dataset and convert it into a Pandas DataFrame.

#### **Question 2:**
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.

    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))

    # Counter for subplot index
    plot_number = 1

    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)

            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]],
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)

            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')

            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

#### **Question 3:**
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

> K-means clustering is highly sensitive to feature magnitudes because it depends on distance computations. If some features span much larger numeric ranges than others, they can disproportionately influence those distances. By standardizing the dataset, all features are placed on a comparable (simialr) scale, preventing any single dimension from dominating the clustering process. Without standardization, the algorithm may produce misleading groupings: subtle patterns in low-scale features can be overshadowed, clusters may form primarily around high-magnitude variables, and interpretations may reflect differences in scale rather than the true underlying structure of the data.

#### **Question 4:**
#### K-means clustering

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# Your answer...

kmeans = KMeans(n_clusters=3, random_state=123, n_init=10)
#fitting the model
kmeans.fit(scaled_wine_df)
#assign cluster labels
cluster_labels = kmeans.labels_

clustered_wine_data = wine_df.copy()
#adding a new column
clustered_wine_data["Cluster"] = cluster_labels

print(clustered_wine_data.head())


# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

**Question**: What is one method commonly used to determine the optimal number of clusters in K-means clustering, and why is this method helpful?

> One commonly used method for determining the optimal number of clusters in K-means is the Elbow Method. This approach involves running K-means for a range of values of k and computing the within-cluster sum of squares (WCSS) for each value. The WCSS typically decreases as k increases, but after a certain point, the rate of improvement drops off. When plotted, this creates a curve with a noticeable “elbow.”

>The location of this elbow represents a good choice for the number of clusters because it indicates a balance between minimizing cluster variance and avoiding unnecessary complexity. For the wine dataset, using the Elbow Method is especially helpful because the features vary in scale and contain inherent group structure (e.g., chemical properties associated with different wine cultivars).

#### **Question 5:**
#### Bootstrapping

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
# Your answer here...

color_intensity = wine_df['color_intensity'].values
n_bootstraps = 10000
mean_color_intensity = []

np.random.seed(123)
for _ in range(n_bootstraps):
    sample = np.random.choice(color_intensity, size=len(color_intensity), replace=True)
    mean_color_intensity.append(np.mean(sample))

mean_color_intensity = np.array(mean_color_intensity)

lower_bound, upper_bound = np.percentile(mean_color_intensity, [5, 95])


# Display the result
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

>  Bootstrapping lets us understand not only the average color_intensity but also how dependable that average is. It shows how much the mean could vary if we repeated the study many times, giving us a better sense of the precision and confidence in our estimate.

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

> The confidence interval created from the bootstrap samples shows how much the sample mean could change if we repeated the study many times. It gives us a range of values that is likely to include the true population mean, helping us understand how reliable and precise our estimate is. A 90% confidence interval means that this range has a high chance of containing the actual population mean.

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

> When we created 10,000 bootstrap samples, the resulting means formed a distribution centered close to the original sample mean. The distribution wasn’t very spread out, meaning most resampled means were similar to each other. The 90% confidence interval was also fairly narrow, indicating a precise estimate. Overall, the tight bootstrap distribution and narrow CI suggest that the mean color_intensity is a stable and reliable estimate of the true population mean.
